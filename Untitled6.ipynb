```python
import gradio as gr

def chatbot_response(message, history):

    message = message.lower()

    if "hello" in message or "hi" in message:
        return "Hello! 👋 I am your Campus Event AI Assistant. How can I help you?"

    elif "upcoming" in message:
        return "You can check the Events page to see upcoming college events."

    elif "event" in message:
        return "You can view all college events from the Events section."

    elif "register" in message:
        return "To register, select an event and click the Register button."

    elif "venue" in message or "location" in message:
        return "The venue of each event is available in the event details."

    elif "date" in message or "time" in message:
        return "The date and time of each event are displayed in the event details."

    elif "recommend" in message:
        return "I can recommend events based on your interests and department."

    elif "thank" in message:
        return "You're welcome! 😊"

    elif "bye" in message:
        return "Goodbye! Have a great day! 👋"

    else:
        return "Sorry, I didn't understand. You can ask about events, registration, dates, venues, or recommendations."


demo = gr.ChatInterface(
    fn=chatbot_response,
    title="🤖 Campus Event AI Assistant",
    description="Ask me about college events"
)

demo.launch(share=True)
```
